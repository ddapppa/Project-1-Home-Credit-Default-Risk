Menggabungkan 6 tabel tambahan (bureau, bureau_balance, previous_application,
POS_CASH_balance, credit_card_balance, installments_payments) ke dataset utama
lewat agregasi per SK_ID_CURR. Skema Opsi A: agregasi seragam sebagai baseline.

# Import

```python
import sys
from pathlib import Path

sys.path.insert(0, str(Path.cwd().parent / "src"))

from modeling.train import load_featured_dataset
from cleaning_auxiliary import load_all_auxiliary_tables
from features_v2 import (
    build_auxiliary_features_v2, merge_with_main_dataset,
    save_featured_dataset_v2, save_feature_metadata_v2,
)
```

## Load Dataset Utama (v1) dan 6 Tabel Tambahan

```python
df_main = load_featured_dataset()
auxiliary_tables = load_all_auxiliary_tables()

print("Shape dataset utama (v1):", df_main.shape)
for name, df in auxiliary_tables.items():
    print(f"{name}: {df.shape}")
```

## Bangun Fitur Agregat dari 6 Tabel (Opsi A)

Proses ini memakan waktu beberapa menit karena `bureau_balance.csv` (27 juta baris)
dan `installments_payments.csv` (13 juta baris) cukup besar.

```python
features_v2 = build_auxiliary_features_v2(auxiliary_tables)
print("Shape fitur v2 (sebelum digabung):", features_v2.shape)
features_v2.head()
```

## Gabungkan ke Dataset Utama dan Simpan

```python
df_v2 = merge_with_main_dataset(df_main, features_v2)
print("Shape dataset utama v1:", df_main.shape)
print("Shape dataset v2 (setelah digabung):", df_v2.shape)

dataset_path = save_featured_dataset_v2(df_v2)
metadata_path = save_feature_metadata_v2(df_v2)

print("Dataset v2 tersimpan di :", dataset_path)
print("Metadata v2 tersimpan di:", metadata_path)
```
